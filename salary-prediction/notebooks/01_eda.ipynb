## Figures
All figures are saved to `reports/figures/`.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import pathlib

figs_dir = pathlib.Path('../reports/figures')
fig_names = sorted(figs_dir.glob('*.png'))

for fp in fig_names:
    img = mpimg.imread(fp)
    plt.figure(figsize=(10, 6))
    plt.imshow(img)
    plt.axis('off')
    plt.title(fp.stem.replace('_', ' ').title())
    plt.tight_layout()
    plt.show()
```
